```python
import torch
import time
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output
```

```python
numOfNodes = 9
numOfLoads = 3
x = np.random.rand(numOfNodes)
y = np.random.rand(numOfNodes)
y[-numOfLoads:] = numOfLoads*[1]
x[0], y[0] = 0, 0
x[1], y[1] = 1,0
plt.scatter(x,y)
centerIndex =  np.sqrt((x-0.5)**2+(y-0.5)**2).argsort()[0]
nodeIndices = np.arctan2(y-y[centerIndex],x-x[centerIndex]).argsort()
for n in range(len(nodeIndices)):
    plt.plot([x[centerIndex],x[nodeIndices[n]]],[y[centerIndex],y[nodeIndices[n]]])
    plt.plot([x[nodeIndices[n-1]],x[nodeIndices[n]]],[y[nodeIndices[n-1]],y[nodeIndices[n]]])
        
```
